```python
config_file = '/kaggle/working/YOLOP/lib/config/default.py'
dataset_root = '/kaggle/input/datasets/eoiwgjlm/bdd45k/yolop_subset'

with open(config_file, 'r') as f:
    content = f.read()

# Reset về đúng giá trị
import re

content = re.sub(r"_C\.DATASET\.DATAROOT\s*=.*", 
    f"_C.DATASET.DATAROOT = '{dataset_root}/images'", content)
content = re.sub(r"_C\.DATASET\.LABELROOT\s*=.*", 
    f"_C.DATASET.LABELROOT = '{dataset_root}/det_annotations'", content)
content = re.sub(r"_C\.DATASET\.MASKROOT\s*=.*", 
    f"_C.DATASET.MASKROOT = '{dataset_root}/da_seg_annotations'", content)
content = re.sub(r"_C\.DATASET\.LANEROOT\s*=.*", 
    f"_C.DATASET.LANEROOT = '{dataset_root}/ll_seg_annotations'", content)
content = re.sub(r"_C\.DATASET\.TRAIN_SET\s*=.*", 
    "_C.DATASET.TRAIN_SET = 'train'", content)
content = re.sub(r"_C\.DATASET\.TEST_SET\s*=.*", 
    "_C.DATASET.TEST_SET = 'val'", content)

with open(config_file, 'w') as f:
    f.write(content)

# Kiểm tra
!grep -n "DATAROOT\|LABELROOT\|MASKROOT\|LANEROOT\|TRAIN_SET\|TEST_SET" \
    /kaggle/working/YOLOP/lib/config/default.py
```

```python
with open('/kaggle/working/YOLOP/tools/test.py', 'r') as f:
    content = f.read()

content = content.replace(
    'checkpoint_file = args.weights',
    'checkpoint_file = args.weights[0] if isinstance(args.weights, list) else args.weights'
)

with open('/kaggle/working/YOLOP/tools/test.py', 'w') as f:
    f.write(content)
print("✅ test.py patched")
```

```python
with open('/kaggle/working/YOLOP/lib/core/postprocess.py', 'r') as f:
    content = f.read()

content = content.replace(
    'indices.append((b, a, gj.clamp_(0, gain[3] - 1), gi.clamp_(0, gain[2] - 1)))',
    'indices.append((b, a, gj.long().clamp_(0, int(gain[3]) - 1), gi.long().clamp_(0, int(gain[2]) - 1)))'
)

with open('/kaggle/working/YOLOP/lib/core/postprocess.py', 'w') as f:
    f.write(content)

print("✅ postprocess.py patched")
```
